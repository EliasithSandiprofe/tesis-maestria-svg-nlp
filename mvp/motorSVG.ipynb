```python
import os

class SVGEngine:
    def __init__(self):
        # Renderizador modular del archivo vectorial SVG
        self.template = """
        <svg width="400" height="400" viewBox="0 0 400 400" xmlns="http://www.w3.org/2000/svg">
            <!-- Cuerpo Base de la Camiseta -->
            <path d="M100,50 L300,50 L350,100 L350,150 L300,150 L300,350 L100,350 L100,150 L50,150 L50,100 Z"
                  fill="{color_camiseta}" stroke="#1e272e" stroke-width="3"/>

            <!-- Contenedor del Gráfico Central -->
            <g transform="{transformacion_posicion}">
                <circle cx="200" cy="180" r="48" fill="{color_grafico}" opacity="0.85"/>
                <text x="200" y="185" font-family="Arial" font-size="11" font-weight="bold" text-anchor="middle" fill="white">
                    {elemento_grafico}
                </text>
            </g>

            <!-- Etiquetas de validación visual del MVP -->
            <text x="200" y="315" font-family="Impact" font-size="16" font-weight="normal" text-anchor="middle" fill="{color_texto_inf}" letter-spacing="1">
                ESTILO: {estilo_diseno}
            </text>
            <text x="200" y="338" font-family="Helvetica" font-size="10" text-anchor="middle" fill="#7f8c8d">
                Posición: {texto_posicion}
            </text>
        </svg>
        """

    def renderizar_lienzo(self, color, estilo, elemento, posicion_matrix, posicion_txt, color_grafico):
        # Control de contraste para el texto inferior
        color_texto_inf = "#2c3e50" if color in ["#ffffff", "#f1c40f"] else "#ffffff"
        
        render_data = {
            "color_camiseta": color,
            "estilo_diseno": estilo,
            "elemento_grafico": elemento,
            "transformacion_posicion": posicion_matrix,
            "texto_posicion": posicion_txt,
            "color_grafico": color_grafico,
            "color_texto_inf": color_texto_inf
        }
        
        svg_content = self.template.format(**render_data)
        ruta_salida = os.path.join("static", "mocks", "resultado_tesis.svg")
        
        os.makedirs(os.path.dirname(ruta_salida), exist_ok=True)
        with open(ruta_salida, "w", encoding="utf-8") as f:
            f.write(svg_content)
            
        return ruta_salida

def generate_and_save_svg(id_color, id_estilo, id_elemento, id_posicion):
    """
    TRADUCCIÓN MATEMÁTICA ALINEADA CON TU MODELO DE TESIS
    """
    # 🎨 Tarea 1: Color (0 = Azul, 4 = Negro/Oscuro según tus logs)
    map_color = {
        0: "#2980b9", # Azul (Activado con tu segundo prompt)
        1: "#c0392b", # Rojo
        2: "#27ae60", # Verde
        3: "#f1c40f", # Amarillo
        4: "#1e272e", # Negro / Oscuro (Activado con tu primer prompt)
        5: "#ffffff"  # Blanco
    }
    
    # 🏛️ Tarea 2: Estilo (4 = Vintage según tus logs)
    map_estilo = {
        0: "MINIMALISTA",
        1: "MODERNO",
        2: "FUTURISTA",
        3: "URBANO",
        4: "VINTAGE"   # <-- CORREGIDO: Mapeado al ID 4
    }

    # 🎸 Tarea 3: Elemento Gráfico (3 = Guitarra según tus logs)
    map_elemento = {
        0: "Círculo",
        1: "Cuadrado",
        2: "Triángulo",  
        3: "Guitarra",   # <-- CORREGIDO: Mapeado al ID 3
        4: "Estrella",
        5: "Cráneo",
        6: "Rayo",
        7: "Planeta",
        8: "Águila",
        9: "Logotipo"
    }

    # 📍 Tarea 4: Posición (0 = Centrado según tus logs)
    map_posicion = {
        0: {"trans": "translate(0, 0)", "text": "Centrado"}, # <-- Mapeado al ID 0
        1: {"trans": "translate(-65, -55) scale(0.75)", "text": "Pecho Izquierdo"},
        2: {"trans": "translate(0, -50) scale(0.85)", "text": "Superior"},
        3: {"trans": "translate(0, 30) scale(0.85)", "text": "Inferior"}
    }

    # Extraer valores correspondientes de forma segura
    color_final = map_color.get(id_color, "#2980b9")
    estilo_final = map_estilo.get(id_estilo, "VINTAGE")
    elemento_final = map_elemento.get(id_elemento, "Guitarra")
    pos_data = map_posicion.get(id_posicion, {"trans": "translate(0, 0)", "text": "Centrado"})
    
    # Color contrastante para el círculo del gráfico estampado
    color_grafico = "#f1c40f" if color_final in ["#2980b9", "#1e272e"] else "#c0392b"

    # Construir y guardar el gráfico SVG
    motor = SVGEngine()
    motor.renderizar_lienzo(
        color=color_final,
        estilo=estilo_final,
        elemento=elemento_final,
        posicion_matrix=pos_data["trans"],
        posicion_txt=pos_data["text"],
        color_grafico=color_grafico
    )
    
    return elemento_final, estilo_final
   
```
