```python
import os
from IPython.display import SVG, display

# 1. DEFINICIÓN DEL MOTOR DE GENERACIÓN SVG
class SVGEngine:
    def __init__(self):
        # Plantilla base: Representa la "Camiseta"
        self.template = """
        <svg width="400" height="400" viewBox="0 0 400 400" xmlns="http://www.w3.org/2000/svg">
            <!-- Fondo de la Camiseta -->
            <path d="M100,50 L300,50 L350,100 L350,150 L300,150 L300,350 L100,350 L100,150 L50,150 L50,100 Z"
                  fill="{color_camiseta}" stroke="black" stroke-width="2"/>

            <!-- Espacio para la Figura (Ej: Dragón) -->
            <circle cx="200" cy="180" r="60" fill="{color_figura}" opacity="0.6" />
            <text x="200" y="190" font-family="Arial" font-size="12" text-anchor="middle" fill="white">
                [{figura}]
            </text>

            <!-- Texto Dinámico -->
            <text x="200" y="300" font-family="serif" font-size="24" font-weight="bold"
                  text-anchor="middle" fill="{color_texto}">
                {texto}
            </text>

            <!-- Estilo / Marca de agua -->
            <text x="200" y="330" font-family="Arial" font-size="10" text-anchor="middle" fill="gray">
                Estilo: {estilo}
            </text>
        </svg>
        """

    def generar_svg(self, atributos):
        """
        Transforma el JSON de atributos en un archivo SVG funcional [2].
        """
        # Valores por defecto para evitar errores si el JSON está incompleto
        data = {
            "color_camiseta": atributos.get("color_camiseta", "white"),
            "figura": atributos.get("figura", "sin figura"),
            "color_figura": atributos.get("color_figura", "red"),
            "texto": atributos.get("texto", "DISEÑO"),
            "color_texto": "black",
            "estilo": atributos.get("estilo", "estándar")
        }

        # Generación dinámica del código SVG [2]
        svg_content = self.template.format(**data)

        # Exportación del archivo (ej: camiseta.svg) [2, 4]
        nombre_archivo = "camiseta.svg"
        with open(nombre_archivo, "w") as f:
            f.write(svg_content)

        return nombre_archivo, svg_content

# 2. PRUEBAS CON DATOS SIMULADOS (MOCKS) [1, 3]
# Estos son los ejemplos de JSON extraídos directamente de las fuentes [2, 5]
mock_nlp_output = {
    "color_camiseta": "black",
    "figura": "dragón",
    "color_figura": "crimson",
    "estilo": "japonés",
    "texto": "Samurai"
}

# 3. EJECUCIÓN Y VALIDACIÓN [6]
engine = SVGEngine()
archivo_generado, contenido_visual = engine.generar_svg(mock_nlp_output)

print(f"Éxito: Se ha generado el archivo '{archivo_generado}'")
display(SVG(contenido_visual))
```
